```python
#List Data type
studentInfo=["Amit","Java","MCA",23,80000.00,False]

#len() method
print(len(studentInfo))
```

```python
#append()
marks=[66,76,56,99,87,98]
marks.append(44)
print(marks)

marks.append([33,22])
print(marks)
print(marks[7])
```

```python
#insert()
marks=[66,76,56,99,87,98]
marks.insert(1,33)
print(marks)
```

```python
#extend()
marks=[66,76,56,99,87,98]
marks.extend([77,99,55])
print(marks)
```

```python
#remove()
marks=[66,76,56,99,87,76,98]
marks.remove(76)
print(marks)
marks.remove(1000)  #ValueError: list.remove(x): x not in list
print(marks)
```

```python
#pop()
marks=[66,76,56,99,87,76,98]
marks.pop()
print(marks)
marks.pop(-1)
print(marks)
marks.pop(-11) #IndexError: pop index out of range
print(marks)
```

```python
#clear()
marks=[66,76,56,99,87,76,98]
marks.clear()
print(marks)
```

```python
#index()
marks=[66,76,56,99,87,76,98,76]
print(marks.index(99))
print(marks.index(76))
print(marks.index(76,2))
```

```python
#count()
marks=[66,76,56,99,87,76,98,76]
print(marks.count(76))
```

```python
#sort()
marks=[66,76,56,99,87,76,98,76]
#marks.sort(reverse=False) # A-Z
marks.sort() # A-Z
print(marks)

marks.sort(reverse=True) # Z-A
print(marks)

#sort() vs sorted()
```

```python
# reverse()

marks = [66, 76, 56, 99, 87, 76, 98, 76]

marks.reverse()
print(marks)
```
