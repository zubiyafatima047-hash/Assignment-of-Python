```python
def even_number(start, end):
    for num in range(start, end + 1):
        if num % 2 == 0:
            yield num

ex = even_number(1, 10)

print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex, "Error: iteration Over"))
```

```python
# Student records
students = [
    {"name": "Aman", "marks": [80, 75, 90]},
    {"name": "Riya", "marks": [35, 40, 30]},
    {"name": "Rahul", "marks": [90, 85, 95]},
    {"name": "Neha", "marks": [60, 70, 65]}
]

def generate_result(student_record):
    for student in student_record:
        name = student['name']
        marks = student['marks']

        avg = sum(marks) / len(marks)

        if avg >= 40:
            status = "Pass"
        else:
            status = "Fail"

        yield {
            "name": name,
            "average": avg,
            "status": status
        }

ex = generate_result(students)

print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```

```python
# Student records

students = [
    {"name": "Aman", "marks": [80, 75, 90]},
    {"name": "Riya", "marks": [35, 40, 30]},
    {"name": "Rahul", "marks": [90, 85, 95]},
    {"name": "Neha", "marks": [60, 70, 65]}
]


def generate_result(student_record):
    for student in student_record:
        name = student['name']
        marks = student['marks']

        avg = sum(marks) / len(marks)

        if avg >= 40:
            status = "Pass"
        else:
            status = "Fail"

        yield {
            "name": name,
            "average": avg,
            "status": status
        }


for result in generate_result(students):
    print(result['name'])
    print(result['average'])
    print(result['status'])
    print("-" * 25)
```
