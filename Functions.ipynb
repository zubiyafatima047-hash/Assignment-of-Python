```python
def add(a,b):
    return a+b
def add(a,b,c):
    return a+b+c
def add(a,b,c,d):
    return a+b+c+d
def add(a,b,c,d,e):
    return a+b+c+d+e

add(10,20)
add(20,30,40)
add(20,30,40,50)
add(20,30,40,50,60)
```

```python
# args
def add(*args):
    total=0
    for ex in args:
        total=total+ex
    print("Sum: ",total)
add(10,20)
add(20,30,40)
add(20,30,40,50)
add(20,30,40,50,60,)
```

```python
def add(*args):
    print("args[0]",args[0])
    print("args[1]",args[1])
    print("args[2]",args[2])
    print("args[3]",args[3])
    print("args[4]",args[4])
    print(args)


add(20,30,40,50,60)
```

```python
def calculate_discount(bill, discount=None):

    if bill <= 0:
        print("Invalid Bill Amount")
        return

    if discount is None:
        if bill < 1000:
            discount = 5
        elif bill < 5000:
            discount = 10
        elif bill < 10000:
            discount = 15
        else:
            discount = 20

    if discount < 0 or discount > 100:
        print("Invalid Discount")
        return

    discount_amount = bill * discount / 100
    final_amount = bill - discount_amount

    print("Original Bill:", bill)
    print("Discount Percentage:", discount, "%")
    print("Discount Amount:", discount_amount)
    print("Final Payable Amount:", final_amount)
    print("=======================================")


calculate_discount(800)
calculate_discount(3000)
calculate_discount(6000)
calculate_discount(12000)
calculate_discount(2000, 25)
```

```python
def student_result(name, *marks, **details):
    print("Student Name:", name)
    print("Subject marks: ",marks)

    if len(marks)>0:
       total = sum(marks)
       average = total / len(marks)

    print("Marks:", marks)
    print("Total Marks:", total)
    print("Average Marks:", average)
    print("=============Additional Information=================")
    for key, value in details.items():
        print(key, ":", value)


student_result(
    "Zubiya", 80, 75, 90,
    roll_number=101,
    semester=1
)
```

```python
def calculate_avg(*marks):
    if not marks:
        return "Error: NO Marks Provided"
    for ex in marks:
        if not isinstance(ex,(int,float)) or isinstance(ex,bool):
            return "Error: Marks Must be Numaric"
        if ex<0 or ex>100:
            return "Error: Marks must be between 0 and 100"
    return sum(marks)/len(marks)

print(calculate_avg(55,45,58,97.0,87,67))
print(calculate_avg())
print(calculate_avg(45,76,True))
```
