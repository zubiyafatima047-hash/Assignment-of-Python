```python
#Tuple
    #Important Point: Parentheses () are commonly used to create tuples

data=(10,20,30,40,50)
print(type(data))
```

```python
# But parentheses are not actually mandatory
# The comma , is what makes a tuple, not the parentheses.

data2=10,20,30,40,50
print(type(data2))
```

```python
data3=(10)
print(type(data3))
# Output: <class 'int'>

# Why?
# Because (10) is not a tuple. It is simply an integer 10 enclosed in parentheses.

#Python treats it as:
# data3 = 10
```

```python
# How to create a single-element tuple?
# You must use a comma:

data4 = (10,)
print(type(data4))

# Remember this rule:
# Code          Type
# data = (10)   int
# data = (10,)  tuple
# data = 10,    tuple
# data = (10, 20)  tuple
```

```python
# Tuple Indexing
# Tuples are ordered and immutable.

data = (10, 20, 30, 40, 50)

print(data[2])  # Indexing is used to access individual elements of a tuple and Tuple indexing starts from 0.
print(data[-1])  # Negative indexing starts from -1 from the right side.
```

```python
# A tuple allows duplicate elements.

data = (10, 20, 30, 40, 50, 30)  # Duplicate Elements
print(data)
```

```python
#Tuple is Immutable
#A tuple is immutable, which means its elements cannot be changed, updated, or deleted after creation.

data=(10,20,30,40,50)
data[0]=88 #tries to change the first element from 10 to 88. Therefore, Python raises a TypeError.

#TypeError: 'tuple' object does not support item assignment
```

```python
#Tuple Packing:
studentData="Aman","MCA AI",78.65 #Tuple Packing: Tuple Packing means storing multiple values into a single tuple without explicitly using parentheses.
print(type(studentData))

#Tuple Unpacking
name,course,avgMarks=studentData #Tuple Unpacking means extracting the elements of a tuple into separate variables.
print(name)

#Note: The number of variables should normally match the number of tuple elements.
```

```python
record = "Amit", "MCA", 23, True  # Tuple Packing

print(record[0])
print(record[1])
print(record)

name, course, age, att = record
# Tuple Unpacking

print(name)

# Unpacking Using *
# * in tuple unpacking is called extended unpacking.
# It allows one variable to collect multiple remaining values.

n, *ab = record

# Note: The * operator collects the remaining elements into a list.

print(n)
print(ab)
```

```python
record = "Amit", "MCA", 23, True  # Tuple Packing

name, *ab, att = record

print(name)
print(ab)
print(att)
```

```python
# len() Returns the total number of elements

data = (10, 20, 30, 40, 50, 30)

print(len(data))


# count() : Counts how many times a value occurs

print(data.count(30))


print(sum(data))  # Returns the sum of numeric elements
print(max(data))  # Returns the largest element
print(min(data))  # Returns the smallest element
print(data.index(10))  # Returns the index of the first occurrence
```

```python
x, y = 10, 20

print(x + y)
print(x - y)
print(x * y)
print(x / y)


# Floor Division

print(7 // 2)   # 3
print(-7 // 2)  # -4

# Exponentiation ** is used to calculate the power of a number.

print(2**3)  # 8


# Important Note:
# Floor division does not simply remove the decimal part;
# it moves the result toward negative infinity.
```

```python
fistName = "Amit"
lastName = "Kumar"

print(fistName + lastName)
```

```python
# Compound Assignment Operators:
# Compound assignment operators are used to perform an operation
# and assign the result back to the same variable.

x, y = 40, 50

x += y  # x = x + y >>> x = 40 + 50 >>> x = 90
print(x)

x -= y  # x = x - y >>> x = 90 - 50 >>> x = 40
print(x)

x *= y  # x = x * y >>> x = 40 * 50 >>> x = 2000
print(x)

x /= y  # x = x / y >>> x = 2000 / 50 >>> x = 40.0
print(x)
```

```python
# Relational / Comparison Operators in Python.
# Relational operators are used to compare two values.
# The result of every relational operator is always a Boolean value: True or False

x = 10
y = 20

print(x == y)  # Equal to
print(x > y)   # Greater Than
print(x < y)   # Less Than
print(x >= y)  # Greater Than or Equal To
print(x <= y)  # Less Than or Equal To
print(x != y)  # Not Equal To
```

```python
# Logical AND Operator in Python:
# The and operator is used to combine two or more conditions.
# and = BOTH conditions must be True.

x = 10
y = 20

print(x < y and y > x)
print(x < y and x == y)
print(x > y and x < y)
print(x > y and x == y)
```
