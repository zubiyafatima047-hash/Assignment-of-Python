```python

# Exercise 6: Movie Ticket Booking

seats = int(input("Enter Available Seats: "))

if seats > 0:
    age = int(input("Enter Customer Age: "))
    price = 200

    if age < 5:
        print("Ticket is Free")
        print("Ticket Price: 0")
    elif age <= 12:
        discount = price * 50 / 100
        finalPrice = price - discount
        print("50% Discount")
        print("Ticket Price:", finalPrice)
    elif age > 60:
        discount = price * 30 / 100
        finalPrice = price - discount
        print("30% Discount")
        print("Ticket Price:", finalPrice)
    else:
        print("Full Price")
        print("Ticket Price:", price)

    print("Ticket Booked Successfully")
else:
    print("House Full")
```

```python

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
elif units <= 300:
    bill = units * 10
else:
    bill = units * 15

if bill > 3000:
    bill = bill + (bill * 10 / 100)

print("Electricity Bill =", bill)
```
